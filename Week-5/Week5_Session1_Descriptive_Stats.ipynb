## 0. Mission and the idea of a summary

Managers often ask for "the average." Sometimes that is a fair picture. Sometimes a few unusually long deliveries pull the average up, and most customers waited less than that number.

A **distribution** is the full pattern of values: the middle, the spread, and the unusual tail.

A **population** is the whole group we want to talk about. A **sample** is the subset we actually measured. Today, this file will play the role of the population. We will then draw small samples from it, so we can see how unstable a small-sample average can be.

A **probability**, in the way we use it today, is a share: if we pick one known delivery at random, how often would it be later than 45 minutes?

## 2. Descriptive statistics

Start with a few numbers, then say them in ordinary words.

```
print("Mean minutes:", round(waits["minutes"].mean(), 2))
```

```
print("Median minutes:", waits["minutes"].median())
```

```
print("Standard deviation:", round(waits["minutes"].std(), 2))
```

```
print("Minimum:", waits["minutes"].min())
```

```
print("Maximum:", waits["minutes"].max())
```

```
waits["minutes"].describe()
```

**How to say this:**
- The **mean** is the centre of mass: add all times and divide by how many there are. Here it is about 32.8 minutes.
- The **median** is the middle value once the times are ordered. Here it is 27.85 minutes. Half of the deliveries are at or below that.
- The **standard deviation** describes typical spread around the mean. A large standard deviation means the times are widely scattered.
- The minimum is about 8 minutes. The maximum is 132 minutes, a very long delay.

The mean is **higher** than the median. That usually means a right-skewed shape: a tail of long delays pulling the average up. Most deliveries are closer to the median than to the mean.

If a manager only heard "32.8 minutes on average," they would have a gloomier typical picture than most customers actually experienced. The mean is not wrong. It is incomplete.

## 3. Distributions

A histogram makes the shape visible.

**Question:** Are QuickDrop times bunched together, or do they have a long tail of delays?

```
plt.figure(figsize=(7, 4))
```

```
plt.hist(waits["minutes"], bins=15)
```

```
plt.axvline(waits["minutes"].mean())
```

```
plt.axvline(waits["minutes"].median(), linestyle="--")
```

```
plt.title("Delivery times are right-skewed: a long tail pulls the mean above the median")
```

```
plt.xlabel("Minutes")
```

```
plt.ylabel("Number of deliveries")
```

```
plt.tight_layout()
```

```
plt.show()
```

The solid line is the mean. The dashed line is the median. Most of the mass sits to the left, with a tail to the right. That is why the two summaries disagree.

When someone asks for "the average," it is fair to answer with both numbers, and with a sentence about the tail.

## 5. Sampling

For this class, the 118 known times are our **population**. Now we draw small **samples** and watch the sample mean move.

A sample mean is an estimate. With only 12 deliveries, that estimate can land quite far from the true mean.

```
pop_mean = waits["minutes"].mean()
```

```
print("Population mean:", round(pop_mean, 2))
```

```
sample1 = waits.sample(n=12, random_state=1)
```

```
print("Sample 1 mean (n=12):", round(sample1["minutes"].mean(), 2))
```

```
sample2 = waits.sample(n=12, random_state=2)
```

```
print("Sample 2 mean (n=12):", round(sample2["minutes"].mean(), 2))
```

```
sample3 = waits.sample(n=12, random_state=3)
```

```
print("Sample 3 mean (n=12):", round(sample3["minutes"].mean(), 2))
```

**Interpretation:** the true mean is about 32.8 minutes. Three unlucky or lucky samples of 12 can come out around the low 20s or the high 30s. A manager who timed only a dozen deliveries could honestly report a number that is quite far from the whole operation.

Larger samples tend to land closer to the population mean, but they are not magic. Next session we will talk about confidence intervals as a way to show that uncertainty.

`random_state` is just a way to get the same sample again when we re-run the notebook. It is not a statistical method.

## 6. Probability as a share

**Question:** If we pick one delivery with a known time at random, what is the probability that it took more than 45 minutes?

```
late = waits[waits["minutes"] > 45]
```

```
print("Late deliveries:", len(late))
```

```
print("Known deliveries:", len(waits))
```

```
print("P(minutes > 45):", round(len(late) / len(waits), 3))
```

**How to say this:** about 18.6% of known deliveries took more than 45 minutes. If we picked one of these 118 deliveries at random, the probability it is over 45 minutes is 0.186.

This is a share in a finished list, not a forecast about next month, and not a claim about every city. It also ignores the two deliveries with a missing time.

## Recap
- A mean is one summary, not the whole story. Compare it with the median and look at the shape.
- Right skew (a tail of large values) pulls the mean above the median.
- A **population** is the group you want to talk about; a **sample** is what you measured. Small samples can miss the population mean by a lot.
- A simple probability can be a share of cases, with a clearly stated base.
- Say what a number means, and what it does not mean.
