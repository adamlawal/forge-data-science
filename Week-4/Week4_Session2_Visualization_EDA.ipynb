## 0. Mission and chart choice

A chart is a sentence in pictures. If you cannot say the sentence, do not draw the picture yet.

### A simple chooser
| If your question is about... | Start with |
|---|---|
| A number compared across groups (which day is highest?) | **Bar** |
| The shape of one numeric column (are most tips small?) | **Histogram** |
| How two numbers move together | **Scatter** |
| How a number changes across an ordered sequence (days in week order) | **Line** |
| The spread of a number in each group, including unusual values | **Box** |
| Several numeric relationships at a glance | **Heatmap** |

You will also use tables (`groupby`) alongside charts. A chart without a count can mislead.

**Attribution:** `tips.csv` comes from the public tips dataset often distributed with seaborn, based on a food server's tips (Bryant and Smith, *Practical Data Analysis: Case Studies in Business Statistics*, 1995).

## 2. Bar chart

**Question:** On which day of the week is the average tip highest?

```
mean_tip_day = tips.groupby("day")["tip"].mean()
```

```
day_order = ["Thur", "Fri", "Sat", "Sun"]
```

```
mean_tip_day = mean_tip_day.reindex(day_order)
```

```
mean_tip_day
```

```
plt.figure(figsize=(7, 4))
```

```
plt.bar(mean_tip_day.index.astype(str), mean_tip_day.values)
```

```
plt.title("Average tip is highest on Sunday")
```

```
plt.xlabel("Day")
```

```
plt.ylabel("Average tip")
```

```
plt.tight_layout()
```

```
plt.show()
```

```
tips.groupby("day")["tip"].agg(["mean", "median", "count"]).round(3)
```

**Interpretation:** Sunday has the highest average tip, about 3.26, ahead of Saturday. Friday looks slightly lower, but Friday is based on only 19 bills, so we should not make a strong Friday claim.

The title already states the takeaway. The table next to the chart keeps us honest about counts.

## 3. Histogram

**Question:** What does the shape of tip amounts look like? Are most tips clustered, or stretched out?

```
plt.figure(figsize=(7, 4))
```

```
plt.hist(tips["tip"], bins=15)
```

```
plt.title("Most tips fall between about 1.50 and 4.00")
```

```
plt.xlabel("Tip amount")
```

```
plt.ylabel("Number of bills")
```

```
plt.tight_layout()
```

```
plt.show()
```

```
print("Mean tip:", round(tips["tip"].mean(), 3))
```

```
print("Median tip:", tips["tip"].median())
```

**Interpretation:** Tips pile up around 2 to 3, with a tail of larger tips. The mean (about 3.00) sits close to the median (2.90), so a few large tips pull the mean only a little. A histogram is the right tool when the question is about shape, not about comparing groups.

## 4. Scatter plot

**Question:** Do larger bills tend to come with larger tips?

```
plt.figure(figsize=(7, 4))
```

```
plt.scatter(tips["total_bill"], tips["tip"])
```

```
plt.title("Larger bills tend to come with larger tips")
```

```
plt.xlabel("Total bill")
```

```
plt.ylabel("Tip")
```

```
plt.tight_layout()
```

```
plt.show()
```

```
print("Correlation:", round(tips["tip"].corr(tips["total_bill"]), 3))
```

**Interpretation:** The points trend upward. The correlation is about 0.68, a moderately strong positive relationship. That is not a rule for every table: some large bills still have modest tips. Correlation is not a reason to say the bill *causes* the tip; it only describes how the two numbers move together.

## 5. Line chart

**Question:** If we walk through the week in order, how does the average tip move from Thursday to Sunday?

```
plt.figure(figsize=(7, 4))
```

```
plt.plot(mean_tip_day.index.astype(str), mean_tip_day.values, marker="o")
```

```
plt.title("Average tip rises toward the weekend, peaking on Sunday")
```

```
plt.xlabel("Day")
```

```
plt.ylabel("Average tip")
```

```
plt.tight_layout()
```

```
plt.show()
```

**Interpretation:** We already knew Sunday was highest. The line is useful because the days are in week order. If we had left the days in alphabetical order, Friday would have come first and the story would have looked wrong.

A line chart is for something that has a natural order (time, weekdays, size). It is not automatically better than a bar.

## 6. Box plot

**Question:** How do lunch and dinner tips compare, not just in their averages, but in their spread?

```
plt.figure(figsize=(7, 4))
```

```
sns.boxplot(data=tips, x="time", y="tip")
```

```
plt.title("Dinner tips are a little higher and more spread out than lunch tips")
```

```
plt.xlabel("Time of day")
```

```
plt.ylabel("Tip")
```

```
plt.tight_layout()
```

```
plt.show()
```

```
tips.groupby("time")["tip"].agg(["mean", "median", "count"]).round(3)
```

**How to read a box:** the box shows the middle half of the values. The line inside is the median. The whiskers reach toward the more extreme typical values. Points beyond the whiskers are unusual values, not automatically errors.

**Interpretation:** Dinner has a higher median and more high-end tips than lunch. Dinner also has many more bills (176 vs 68), so the dinner picture is more stable.

## 7. Heatmap

**Question:** How strongly are bill, tip, and party size related to one another?

```
num_cols = tips[["total_bill", "tip", "size"]]
```

```
corr = num_cols.corr()
```

```
corr
```

```
plt.figure(figsize=(5, 4))
```

```
sns.heatmap(corr, annot=True, fmt=".2f", cmap="vlag", vmin=-1, vmax=1)
```

```
plt.title("Tip is more closely related to bill size than to party size")
```

```
plt.tight_layout()
```

```
plt.show()
```

**Interpretation:** tip and total_bill have the strongest relationship (about 0.68). Party size is also positively related to both, but more weakly. A heatmap is a compact map of several correlations. It does not tell us what to change in the restaurant, and it does not prove cause.

## Recap
- Every chart needs a **question**, a **title that states a takeaway**, labelled axes, and a short interpretation.
- Bar: compare numbers across groups. Hist: see the shape of one column. Scatter: two numbers together. Line: an ordered sequence. Box: spread by group. Heatmap: several correlations at once.
- Always read the **counts**. Small groups (Friday, Thursday dinner) should soften the claim.
- A relationship in a scatter or heatmap is not automatically a cause.
- The EDA project uses this same standard: charts without interpretation do not pass.
