## 1. A worked mini example

To make the shape concrete, we will walk a small, complete question on `mini_capstone_orders.csv`. These are the completed orders from Sunrise Stores, the shop we used in Week 6, already aggregated to **one row per completed order**.

**Problem.** Should Sunrise Stores add extra Saturday staff, on the assumption that weekend orders bring in more money than weekday orders?

That is a decision-shaped question. We will not have staffing-cost data, so we cannot answer the full business decision. We *can* check the assumption underneath it: are weekend baskets actually larger?

**Dataset.** 358 completed orders from January to October 2026, with order date, weekday, a weekend flag, channel, city, customer segment, number of items, and revenue.


````python
import pandas as pd
````

````python
import matplotlib.pyplot as plt
````

````python
orders = pd.read_csv("mini_capstone_orders.csv")
````

````python
print(orders.shape)
````

````python
orders.isna().sum()
````

````python
print(orders["is_weekend"].value_counts())
````

````python
print(orders["is_weekend"].mean())
````

There should be 358 rows and no missing values. `is_weekend` is 1 for Saturday and Sunday. The weekend share should be close to 28 percent, which is roughly two days out of seven. So the *number* of weekend orders is about what you would expect from the calendar, not a special weekend rush.


**INSTRUCTOR NOTE.** 358 rows, 9 columns, no missing values. Weekend: 101 rows (0.282). Weekday: 257. If someone says "there are fewer weekend orders, so weekends are quiet", remind them there are also fewer weekend days. The fair first check is the share against 2/7, then the average basket.


## 2. Compare weekend and weekday revenue


````python
weekend = orders[orders["is_weekend"] == 1]
````

````python
weekday = orders[orders["is_weekend"] == 0]
````

````python
print(len(weekend))
````

````python
print(len(weekday))
````

````python
print(weekend["revenue"].mean())
````

````python
print(weekday["revenue"].mean())
````

````python
print(weekend["revenue"].median())
````

````python
print(weekday["revenue"].median())
````

````python
print(weekend["revenue"].sum())
````

````python
print(weekday["revenue"].sum())
````

A compact table is easier to read than a stack of print-outs. `groupby` gives us the same numbers in one place.


````python
summary = (
    orders.groupby("is_weekend")["revenue"]
    .agg(["count", "mean", "median", "sum"])
    .round(1)
)
````

````python
summary
````

And the mean revenue by day of the week, in calendar order:


````python
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
by_day = orders.groupby("weekday")["revenue"].agg(["count", "mean"]).reindex(day_order).round(1)
````

````python
by_day
````

The assumption we were testing does not hold. Weekend completed orders have a **lower** average revenue (about 37,200 against about 48,200 on weekdays). The median tells the same story (19,600 against 26,100). Sunday is the softest day; Monday is the strongest.

That is already an insight: if the shop adds Saturday staff because it believes weekend baskets are larger, the belief is not supported by this table.


**INSTRUCTOR NOTE.** Means: weekend 37,198, weekday 48,240, gap about 11,042 the other way from the hypothesis. Medians 19,600 vs 26,100. Totals 3,757,000 vs 12,397,700. Day means: Mon 50,687; Tue 46,195; Wed 49,733; Thu 49,618; Fri 44,558; Sat 38,614; Sun 35,810. Pause and let someone say the finding in one sentence before you plot it.


### Your turn 1

Compute the mean revenue for weekend and weekday orders *without* wholesale customers, and print both means plus the two group sizes. Then, in a sentence, say whether the weekend gap disappears once wholesale orders are removed.


````python
# SOLUTION
non_wholesale = orders[orders["segment"] != "Wholesale"]
weekend_nw = non_wholesale[non_wholesale["is_weekend"] == 1]
weekday_nw = non_wholesale[non_wholesale["is_weekend"] == 0]
print(len(weekend_nw))
print(len(weekday_nw))
print(weekend_nw["revenue"].mean())
print(weekday_nw["revenue"].mean())
````

**SOLUTION — sentence.** The gap shrinks but does not disappear: without wholesale, weekend mean revenue is about 14,771 against about 17,185 on weekdays (65 weekend orders, 151 weekday orders). Wholesale mix explains part of the original gap, not all of it.

**INSTRUCTOR NOTE. Your turn 1.** Wholesale is 41.2 percent of weekday orders and 35.6 percent of weekend orders, and wholesale weekday means are higher (92,478 vs 77,692). If a student stops at "wholesale is the whole story", ask them to look at the non-wholesale means.


## 3. Two charts, each with a takeaway title


````python
plt.figure()
orders.groupby("weekday")["revenue"].mean().reindex(day_order).plot(kind="bar")
plt.ylabel("Mean completed-order revenue")
plt.title("Weekend days have a lower average completed-order value than weekdays")
plt.tight_layout()
plt.show()
````

````python
plt.figure()
orders.groupby("is_weekend")["revenue"].mean().plot(kind="bar")
plt.xticks([0, 1], ["Weekday", "Weekend"])
plt.ylabel("Mean completed-order revenue")
plt.title("Completed weekend baskets average about 11,000 less than weekday baskets")
plt.tight_layout()
plt.show()
````

Each title states the takeaway, not just the chart type. That is the same standard you used in the EDA week, and it still holds for the capstone.


## 4. Findings, recommendation, limitations, future work

**Findings.**

1. Weekend orders are 101 of 358 completed orders, or 28.2 percent, close to two days in seven. The shop is not seeing a special weekend surge in order *count*.
2. Mean completed-order revenue is about 37,198 on weekends and about 48,240 on weekdays. The median gap points the same way (19,600 against 26,100).
3. Wholesale orders are a bit more common on weekdays and have a higher weekday average, so they widen the gap. After removing wholesale, weekend orders are still smaller on average (about 14,771 against 17,185).

**Recommendation.** Do not add Saturday staff on the belief that weekend baskets are larger. They are not, in this window. If the shop still wants to change weekend staffing, it needs a different reason (for example, coverage, or a different service goal), and it needs cost and footfall data that this table does not have.

**Limitations.** This file contains completed orders only, for one shop, from January to October 2026. We do not have hours of staff time, wages, or footfall. We cannot see cancelled baskets here. Segment mix is a confounder, and we only did a simple split, not a formal test. Average revenue per order is not the same as profit.

**Future work.** Bring in staffing hours and cost; look at busy hours inside each day; and, if useful, test whether the weekend-weekday difference still stands after a simple statistical check.
