```python
import pandas as pd
```

**Minute 0 to 1 — the question.** Should Sunrise Stores add extra Saturday staff, on the belief that weekend orders bring in more money?

**Minute 1 to 2 — the data.** 358 completed orders, January to October 2026, one row per completed order. We do not have staffing costs or footfall in this file, so we will test the belief, not the full hiring decision.

**Minute 2 to 5 — the findings.** Weekend orders are 101 of 358, about 28 percent, close to two days in seven, so this is not a weekend surge in count. Mean basket value is about 37,200 on weekends and about 48,200 on weekdays. The median agrees. Removing wholesale shrinks the gap but does not remove it.

**Minute 5 to 7 — recommendation and limitation.** Do not add Saturday staff *because weekend baskets are larger* — they are not. A different reason might still exist, but it would need cost and footfall data we do not have. This is one shop and ten months of completed orders only.

**Minute 7 to 8 — close.** The number to remember is the average basket: about 11,000 lower on the weekend, not higher. Next step, if the shop wants to go further, is to bring in hours and cost.

Notice how little of that outline is a demo of code. The four printed numbers are there to be said aloud, not to be hidden in a long scroll.
