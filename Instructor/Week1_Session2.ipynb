## One-hour preparation route

1. **10 minutes — transactions:** run the first three code cells. Predict the total before checking it. Changing one transaction type changes the transfer count but leaves the amount total unchanged.
2. **10 minutes — data quality and sampling:** run the customer record and sample cells. Do not guess that age 210 means 21; ask students how they would verify it. In the new balance example, the full average is ₦162,000 and the first-four average is ₦387,500. The random-four result will vary.
3. **10 minutes — scale and lifecycle:** change the transaction rate from 2,000 to 20 per second. Ask students to narrate a transfer's path from app entry to report. Spend little time defining storage terms.
4. **10 minutes — debugging:** rehearse the four deliberate errors and their fixes listed below. Let students read the error messages before giving hints.
5. **15 minutes — simulator:** run the three steps in order. The ₦50,000 withdrawal is declined after the deposit leaves ₦35,000; a ₦30,000 withdrawal succeeds. In the day loop, the ₦45,000 debit is declined and the closing balance is ₦26,000. Changing the opening balance changes that path.
6. **5 minutes — practice and assignment:** prioritise independent Tasks A and B. Tasks C and D can extend faster students. Read the assignment brief and expected totals at the end of this guide.

For each demonstration, use the same rhythm: **ask for a prediction, run one cell, ask students to explain the output, then change one value**. This replaces a long explanation with a visible result.

## Talking points

- Quantitative data supports numerical summaries. Qualitative data can be counted, grouped and interpreted.
- Collection methods have different limitations. A log may omit an event it was not designed to capture; a form may contain missing or inaccurate values.
- A random selection is only as representative as its sampling frame. Choosing randomly from branch visitors cannot represent customers who never visit that branch.
- A data lake, warehouse and mart are common storage patterns, not a fixed sequence every dataset must follow. Metadata describes the data, such as column meanings.
- Volume, velocity, variety, veracity and value help frame the practical difficulties and purpose of large datasets.
- Privacy, security, data quality and access control matter at every lifecycle stage.
- Analysis, analytics and data science overlap. Avoid presenting them as rigidly separate professions.

## Questions to ask

- “If ForgeBank studies only app users, whose experience might be missing?”
- “What should happen after a transaction is flagged for review?”
- “Why would a bank automate some checks across a large transaction stream?”

## Common mistakes

- Using `=` where a comparison needs `==`.
- Omitting a colon or indentation.
- Adding text from `input()` to a number without conversion.
- Treating a negative debit as money entering the account.
- Continuing PIN attempts after a correct entry.

# Solutions
## Task A: Loan eligibility

```python
age = 24
income = 180000
has_loan = False

if age < 18:
    print("Rejected: must be 18 or older")
elif income < 150000:
    print("Rejected: income below ₦150,000")
elif has_loan:
    print("Rejected: existing loan")
else:
    print("Approved")
```

## Task B: Fraud flags

```python
amounts = [5000, 1200000, 30000, 750000, 2500000]
hours   = [14,   10,      2,     23,     3]

for amount, hour in zip(amounts, hours):
    if amount > 1000000 or hour < 4:
        print(f"FLAG: ₦{amount:,} at {hour}:00")
    else:
        print(f"OK: ₦{amount:,} at {hour}:00")
```

## Task C: PIN lock

```python
correct_pin = "4321"
attempts = ["1111", "1234", "4321"]

tries = 0
unlocked = False
while tries < 3 and not unlocked:
    entered = attempts[tries]
    tries = tries + 1
    if entered == correct_pin:
        unlocked = True
        print(f"Access granted on attempt {tries}")
    else:
        print(f"Wrong PIN ({3 - tries} attempts left)")

if not unlocked:
    print("Card locked. Contact ForgeBank.")
```

## Task D: Savings growth

```python
balance = 100000
rate = 0.12
for year in range(1, 6):
    balance = balance + balance * rate
    print(f"Year {year}: ₦{balance:,.2f}")
```

# Assignment Solutions: ForgeBank Statement Analyzer

```python
customer = "  tUNDE   bakare "
opening_balance = 80000
transactions = [150000, -12000, -2500, -45000, -800, 20000, -95000, -1500, -60000, 5000]
descriptions = ["salary", "pos", "airtime", "rent", "airtime", "transfer in", "transfer out", "airtime", "transfer out", "refund"]
```

**Task 1**

```python
parts = customer.split()
clean = " ".join(parts).title()
print(f"Statement for: {clean}")
# Also accept: customer.strip().title(), which leaves a double space inside the name
```

**Task 2**

```python
for amount, desc in zip(transactions, descriptions):
    if amount > 0:
        print(f"CREDIT ₦{amount:,} - {desc.upper()}")
    else:
        print(f"DEBIT  ₦{-amount:,} - {desc.upper()}")
```

**Task 3**

```python
money_in = 0
money_out = 0
for amount in transactions:
    if amount > 0:
        money_in = money_in + amount
    else:
        money_out = money_out + (-amount)
closing = opening_balance + money_in - money_out
print(f"Money in:  ₦{money_in:,}")
print(f"Money out: ₦{money_out:,}")
print(f"Closing:   ₦{closing:,}")
```

**Task 4**

```python
airtime_count = 0
airtime_total = 0
for amount, desc in zip(transactions, descriptions):
    if desc == "airtime":
        airtime_count = airtime_count + 1
        airtime_total = airtime_total + (-amount)
print(f"Airtime purchases: {airtime_count}, total ₦{airtime_total:,}")
```

**Task 5**

```python
largest = 0
for amount in transactions:
    if amount < 0 and -amount > largest:
        largest = -amount
print(f"Largest debit: ₦{largest:,}")
```

**Task 6**

```python
if closing > 50000:
    status = "Healthy"
elif closing >= 10000:
    status = "Watch spending"
else:
    status = "Low balance alert"
print(f"Status: {status}")
```

**Task 7: possible responses**

1. Quantitative: transaction amounts and opening balance. Qualitative: descriptions and customer name.
2. A value may be mistyped, a description omitted, or a name entered with inconsistent spacing.
3. Students are analysing the stored and processed statement, then reporting a summary. Accept a reasoned answer that identifies more than one stage.

**Expected results:** money in ₦175,000; money out ₦216,800; closing balance ₦38,200; status `Watch spending`; three airtime purchases totalling ₦4,800; largest debit ₦95,000.
