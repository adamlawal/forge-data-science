## C. Conditions

Test boundaries in descending order. Testing `score >= 50` first would classify a score of 70 as `C` before Python reaches the `A` branch.

```python
score = 70
if score >= 70:
    grade = "A"
elif score >= 60:
    grade = "B"
elif score >= 50:
    grade = "C"
else:
    grade = "F"
print(grade)
```

## D. Counter and average

```python
scores = [78, 45, 62, 39, 91]
needs_review = 0
total = 0
for score in scores:
    total += score
    if score < 50:
        needs_review += 1
print("Needs review:", needs_review)
print("Average:", total / len(scores))
```

## Core build

The grade boundary and review threshold are separate decisions. A student with 50 earns C and is not below the 50-point review threshold.

```python
scores = [78, 45, 62, 39, 91]
review_threshold = 50
total = 0
needs_review = 0
for score in scores:
    if score >= 70:
        grade = "A"
    elif score >= 60:
        grade = "B"
    elif score >= 50:
        grade = "C"
    else:
        grade = "F"
    if score < review_threshold:
        needs_review += 1
    total += score
    print(f"{score}: {grade}")
print(f"Class average: {total / len(scores):.1f}")
print(f"Needs review: {needs_review}")
```

## Homework example: expense checker

```python
expenses = [2500, 1200, 8000, 3500]
budget = 12000
total = 0
for expense in expenses:
    total += expense
print(f"Total spent: ₦{total:,}")
if total > budget:
    print(f"Over budget by ₦{total - budget:,}")
else:
    print(f"Within budget by ₦{budget - total:,}")
```
