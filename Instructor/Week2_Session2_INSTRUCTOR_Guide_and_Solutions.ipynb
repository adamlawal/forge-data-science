## Common mistakes

| Mistake | Symptom | Fix |
|---|---|---|
| `split(",")` on CSV lines | Wrong field count, names split in two | `csv.reader` / `DictReader` |
| Assuming types | `TypeError` adding strings, wrong sorts | Convert with `int`/`float` inside `try` |
| Reading after the `with` block | `ValueError: I/O operation on closed file` | Read inside the block, or `list(reader)` inside it |
| Wrong column case | `KeyError: 'fare'` | Check `reader.fieldnames` |
| `float("")` on blanks | `ValueError` | Skip or return `None` for blanks |
| Writing to the raw path | Raw data lost | New file name; guard in `save_csv` |
| Forgetting `newline=""` | Blank lines between rows on Windows | Always pass `newline=""` |
| File name typo or wrong folder | `FileNotFoundError` | `os.listdir`, `os.path.exists`, use `RAW_PATH` |

## Debugging clinic answers

1. `FileNotFoundError`: the name is `Titanic-Dataset.csv` (capital D) and it lives in the kagglehub folder. Use `RAW_PATH`.
2. `TypeError: unsupported operand type(s) for +: 'int' and 'str'`: Age is text and some ages are blank. Use `to_number(row["Age"])` and skip `None`.
3. `ValueError: I/O operation on closed file`: `next(reader)` is outside the `with` block. Indent it.
4. `KeyError: 'fare'`: the column is `Fare`. Check `columns`.
5. `ValueError: could not convert string to float: ''`: 177 blank ages. Use `to_number` or `if row["Age"]`.
6. It runs, but if called with `RAW_PATH` it would erase the original data with no undo. Use a different file name and a guard like the one in `save_csv`.

# Solutions

Run the setup cell first.

```python
import os

RAW_PATH = None
try:
    import kagglehub                      # already installed in Colab
    folder = kagglehub.dataset_download("yasserh/titanic-dataset")
    RAW_PATH = os.path.join(folder, "Titanic-Dataset.csv")
except Exception as error:
    print("kagglehub download did not work:", error)

# Fallback: a file you uploaded through the Colab Files panel
if (RAW_PATH is None or not os.path.exists(RAW_PATH)) and os.path.exists("Titanic-Dataset.csv"):
    RAW_PATH = "Titanic-Dataset.csv"

print("Raw file:", RAW_PATH)
print("Exists:", RAW_PATH is not None and os.path.exists(RAW_PATH))


import csv

with open(RAW_PATH, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    columns = reader.fieldnames
    rows = list(reader)

print("Columns:", columns)
print("Rows:", len(rows))


import csv, os, re

PORTS = {"S": "Southampton", "C": "Cherbourg", "Q": "Queenstown"}
TITLE_FIX = {"Mlle": "Miss", "Ms": "Miss", "Mme": "Mrs"}

def load_csv(path):
    """Read a CSV file and return (column_names, list_of_row_dicts)."""
    if not os.path.exists(path):
        raise FileNotFoundError(f"No file at {path}")
    with open(path, newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        return reader.fieldnames, list(reader)

def to_number(text, kind=float):
    """Convert text to a number; return None if it is blank or invalid."""
    try:
        return kind(text)
    except (ValueError, TypeError):
        return None

def extract_title(name):
    match = re.search(r",\s*([^.]+)\.", name)
    if not match:
        return "Unknown"
    title = match.group(1).strip()
    return TITLE_FIX.get(title, title)

def clean_row(row):
    """Return a NEW cleaned dict. The input row is not changed."""
    cabin = row["Cabin"].strip()
    sibsp = to_number(row["SibSp"], int)
    parch = to_number(row["Parch"], int)
    return {
        "passenger_id": to_number(row["PassengerId"], int),
        "survived": to_number(row["Survived"], int),
        "pclass": to_number(row["Pclass"], int),
        "name": " ".join(row["Name"].split()),
        "title": extract_title(row["Name"]),
        "sex": row["Sex"].strip().lower(),
        "age": to_number(row["Age"]),
        "age_missing": row["Age"].strip() == "",
        "family_size": sibsp + parch + 1,
        "fare": to_number(row["Fare"]),
        "deck": cabin[0] if cabin else "Unknown",
        "embarked": PORTS.get(row["Embarked"].strip(), "Unknown"),
    }

def validate_row(clean):
    """Return a list of problems with a cleaned row (empty list = OK)."""
    problems = []
    if clean["survived"] not in (0, 1):
        problems.append("survived must be 0 or 1")
    if clean["pclass"] not in (1, 2, 3):
        problems.append("pclass must be 1, 2 or 3")
    if clean["sex"] not in ("male", "female"):
        problems.append("unexpected sex value")
    if clean["age"] is not None and not (0 < clean["age"] <= 100):
        problems.append("age out of range")
    if clean["fare"] is None or clean["fare"] < 0:
        problems.append("fare missing or negative")
    return problems

def save_csv(rows, path, raw_path):
    """Write rows to path, refusing to overwrite the raw file."""
    if os.path.abspath(path) == os.path.abspath(raw_path):
        raise ValueError("Refusing to overwrite the raw file")
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=list(rows[0].keys()))
        writer.writeheader()
        writer.writerows(rows)

def run_pipeline(raw_path, out_path):
    columns, raw_rows = load_csv(raw_path)
    cleaned, rejected = [], []
    seen_ids = set()
    for line_number, row in enumerate(raw_rows, start=2):   # line 1 is the header
        clean = clean_row(row)
        problems = validate_row(clean)
        if clean["passenger_id"] in seen_ids:
            problems.append("duplicate passenger_id")
        seen_ids.add(clean["passenger_id"])
        if problems:
            rejected.append((line_number, problems))
        else:
            cleaned.append(clean)
    save_csv(cleaned, out_path, raw_path)
    print(f"Read {len(raw_rows)} rows, kept {len(cleaned)}, rejected {len(rejected)}")
    print(f"Saved {out_path}")
    return cleaned, rejected

from collections import Counter
from statistics import mean, median
cleaned, rejected = run_pipeline(RAW_PATH, "cleaned_titanic.csv")
```

```python
def survival_rate(rows):
    if not rows:
        return 0
    return 100 * sum(r["survived"] for r in rows) / len(rows)

for pclass in (1, 2, 3):
    group = [r for r in cleaned if r["pclass"] == pclass]
    print(f"Class {pclass}: {len(group)} passengers, {survival_rate(group):.1f}% survived")
```

# Challenge solution

```python
known_by_title = {}
for r in cleaned:
    if r["age"] is not None:
        known_by_title.setdefault(r["title"], []).append(r["age"])
title_median = {t: median(ages) for t, ages in known_by_title.items()}
overall_median = median(r["age"] for r in cleaned if r["age"] is not None)

estimated_rows = []
used = Counter()
for r in cleaned:
    new = dict(r)                       # copy, so cleaned stays unchanged
    if new["age"] is None:
        new["age"] = title_median.get(new["title"], overall_median)
        new["age_estimated"] = True
        used[new["title"]] += 1
    else:
        new["age_estimated"] = False
    estimated_rows.append(new)

save_csv(estimated_rows, "cleaned_titanic_estimated_ages.csv", RAW_PATH)
print("Ages estimated:", sum(used.values()))
for title, n in used.most_common():
    print(f"{title:<7} {n:>3} filled with median {title_median.get(title, overall_median)}")
print("Original cleaned still has missing ages:", sum(1 for r in cleaned if r["age"] is None))
```

Written part, model answer: filling values invents data and can hide uncertainty or bias averages towards the median; the `age_estimated` column lets anyone exclude or label estimated ages, so the exhibition stays honest. If a title has no known ages (Dr has some; all titles here do), fall back to the overall median.
