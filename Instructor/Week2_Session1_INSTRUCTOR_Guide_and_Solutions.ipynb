## Common mistakes to watch for

| Mistake | Symptom | Fix |
|---|---|---|
| `b = a` instead of `b = a.copy()` | Original list changes | Use `.copy()` or `list(a)` |
| Position lookup on a dict, key lookup on a list | `KeyError: 0` or `TypeError: list indices must be integers` | Lists by position, dicts by key |
| Function without `return` | `None` appears, or `TypeError` with NoneType | Add `return` |
| `x = my_list.sort()` | `x` is `None` | Use `sorted(my_list)` or sort then use `my_list` |
| `("Ikeja")` for a tuple | A string, not a tuple | `("Ikeja",)` |
| Indexing a set | `TypeError: 'set' object is not subscriptable` | Loop over it or convert to a sorted list |
| Looping a dict expecting pairs | `ValueError: too many values to unpack` | `for k, v in d.items()` |
| Bare `except:` | Typos silently ignored | Catch a named exception |
| Missing colon or indentation after `def`, `if`, `for` | `SyntaxError` or `IndentationError` | Colon, then indent four spaces |

## Debugging clinic answers

1. `prices[3]`: `IndexError`; the list has positions 0, 1, 2. Use `prices[2]` or `prices[-1]`.
2. `product[0]`: `KeyError: 0`; dictionaries use keys. Use `product["name"]`.
3. `delivery_fee` has no `return`, so it returns `None`, and `9800 + None` raises `TypeError` on the `total` line, not the function line. Add `return fee`.
4. `quantity` is the string `"4"`. `"4" * 9800` repeats the text 9,800 times, then `+ 1500` raises `TypeError`. Use `int(quantity)`.
5. Sets have no positions: `TypeError: 'set' object is not subscriptable`. Use `sorted(areas_seen)[0]` or loop.
6. Looping over a dictionary gives keys only, so unpacking `"P01"` into two names fails with `ValueError`. Use `stock.items()`.
7. Mutation surprise: `discounted = original_prices` is the same list, so both change. Use `original_prices.copy()` or a comprehension: `[p * 0.9 for p in original_prices]`.

## Task D

```python
stock = {item["sku"]: item["stock"] for item in catalogue}

def sell(stock, sku, qty):
    try:
        current = stock[sku]
    except KeyError:
        print(f"{sku} does not exist")
        return None
    if qty > current:
        print(f"Not enough {sku}: asked for {qty}, only {current} left")
        return current
    stock[sku] = current - qty
    return stock[sku]

print(sell(stock, "P02", 5))
print(sell(stock, "P04", 9))
print(sell(stock, "P99", 1))
print(stock)
```

# Challenge solution: Rider assignment

```python
riders = {
    "Musa":   ("Ikeja", "Yaba"),
    "Funke":  ("Lekki", "Ajah"),
    "Segun":  ("Surulere", "Yaba"),
}

def riders_for(area, riders):
    return [name for name, covered in riders.items() if area in covered]

for o in clean_orders:
    print(o["id"], o["area"], "->", riders_for(o["area"], riders))

order_areas = {o["area"] for o in orders}
covered = set()
for zone in riders.values():
    covered = covered | set(zone)
print("Uncovered areas:", order_areas - covered or "none")
print("Shared areas:", set(riders["Musa"]) & set(riders["Segun"]))

# Stretch: least-busy rider
counts = {name: 0 for name in riders}
for o in clean_orders:
    options = riders_for(o["area"], riders)
    if options:
        chosen = min(options, key=lambda name: counts[name])
        counts[chosen] += 1
print(counts)
```

Using all `orders` instead of `clean_orders` is acceptable; the uncovered result is still none because every order area is covered. A full pairwise check of shared areas is a good extension.

# Assignment solutions: EkoMart Customer Insights

```python
week_orders = [
    {"id": "A01", "customer": " ada okafor",  "area": "Ikeja",    "items": ["P02", "P05"],        "total": "24200", "day": "Mon"},
    {"id": "A02", "customer": "TUNDE BAKARE", "area": "Lekki",    "items": ["P01"],               "total": "139000", "day": "Mon"},
    {"id": "A03", "customer": "chioma eze",   "area": "yaba",     "items": ["P02", "P02", "P04"], "total": "51600", "day": "Tue"},
    {"id": "A04", "customer": "Ada Okafor",   "area": "Ikeja",    "items": ["P03"],               "total": "18500", "day": "Wed"},
    {"id": "A05", "customer": "Emeka Obi",    "area": "Ajah",     "items": [],                    "total": "0",     "day": "Wed"},
    {"id": "A06", "customer": "zainab musa",  "area": "Surulere", "items": ["P05", "P05"],        "total": "28,800", "day": "Thu"},
    {"id": "A07", "customer": "Bola Ade",     "area": "LEKKI",    "items": ["P04", "P02"],        "total": "41800", "day": "Fri"},
    {"id": "A08", "customer": "Tunde Bakare", "area": "Lekki",    "items": ["P04"],               "total": "N/A",   "day": "Sat"},
]
```

**Task 1**

```python
def clean_name(name):
    return name.strip().title()

print([clean_name(o["customer"]) for o in week_orders])
```

**Task 2**

```python
def to_naira(text):
    try:
        return int(text.replace(",", ""))
    except ValueError:
        return None

for o in week_orders:
    print(o["id"], to_naira(o["total"]))
```

**Task 3**

```python
flagged = []
for o in week_orders:
    amount = to_naira(o["total"])
    if not o["items"] or amount is None or amount == 0:
        flagged.append(o["id"])
print("Flagged:", flagged)
```

**Task 4**

```python
from statistics import mean
good = [o for o in week_orders if o["id"] not in flagged]
good_totals = [to_naira(o["total"]) for o in good]
revenue = sum(good_totals)
print(f"Revenue: ₦{revenue:,}")
print(f"Average order: ₦{mean(good_totals):,.0f}")
```

**Task 5**

```python
area_revenue = {}
for o in good:
    area = o["area"].strip().title()
    area_revenue[area] = area_revenue.get(area, 0) + to_naira(o["total"])
for area, amount in sorted(area_revenue.items(), key=lambda pair: pair[1], reverse=True):
    print(f"{area:<9} ₦{amount:,}")
```

**Task 6**

```python
from collections import Counter
names = [clean_name(o["customer"]) for o in week_orders]
unique_customers = set(names)
print("Unique customers:", len(unique_customers))
counts = Counter(names)
print("Ordered more than once:", sorted(n for n, c in counts.items() if c > 1))
```

**Task 7**

```python
sku_counts = {}
for o in week_orders:
    for sku in o["items"]:
        sku_counts[sku] = sku_counts.get(sku, 0) + 1
print(sku_counts)
top_sku = max(sku_counts, key=sku_counts.get)
print("Top SKU:", top_sku, "with", sku_counts[top_sku], "units")
```

**Task 8**

```python
def weekly_report(orders):
    flagged = []
    good_totals = []
    area_revenue = {}
    for o in orders:
        amount = to_naira(o["total"])
        if not o["items"] or amount is None or amount == 0:
            flagged.append(o["id"])
            continue
        good_totals.append(amount)
        area = o["area"].strip().title()
        area_revenue[area] = area_revenue.get(area, 0) + amount
    return {
        "revenue": sum(good_totals),
        "average": round(mean(good_totals), 2) if good_totals else 0,
        "unique_customers": len({clean_name(o["customer"]) for o in orders}),
        "top_area": max(area_revenue, key=area_revenue.get) if area_revenue else None,
        "flagged": flagged,
    }

for key, value in weekly_report(week_orders).items():
    print(f"{key}: {value}")
```

**Expected results:** flagged `A05` and `A08`; revenue ₦303,900 from 6 orders; average ₦50,650; Lekki ₦180,800, Yaba ₦51,600, Ikeja ₦42,700, Surulere ₦28,800; 6 unique customers, with Ada Okafor and Tunde Bakare ordering twice; P02 is the top SKU with 4 units (P05 and P04 have 3 each); top area Lekki.
