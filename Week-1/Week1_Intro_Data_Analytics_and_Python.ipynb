## 2. Data and the workflow

**Data** is recorded information. A score is data; so are the date of the quiz, whether a student was absent, and how the quiz was marked. Data is useful only when we connect it to a clear question.

| Step | Question to ask in our case |
|---|---|
| Define | Are we identifying students who may need support, or deciding who passes? |
| Collect | Where did scores come from? Do we have permission to use them? |
| Check and clean | Are scores missing, duplicated, or outside 0–100? |
| Explore | What is the average? How many are below 50? |
| Explain | What pattern did we find, and how uncertain is it? |
| Act and review | Offer support, then check whether it helped. |

Prediction or machine learning can come later, **if the question and data justify it**. It is not a required step in every data project.

### Analytics and data science overlap

Data analytics usually emphasizes describing and explaining what happened to inform a decision. Data science is a broader practice that can also include prediction and building data products. The labels vary across workplaces; neither field is defined by one tool or by whether it looks only at the past or future.

## 5. Conditions express a decision rule

A condition is either `True` or `False`. Use `if`, `elif` and `else` to choose a path. The colon and indentation show which code belongs to each path. `>=` means “at least”; `==` tests equality.

```python
score = 62
print(score >= 50)
if score >= 50:
    print("Pass")
else:
    print("Needs review")
```

**Important:** A threshold is a rule we choose, not a fact discovered in the data. For today, a score below 50 triggers a *review*, not an automatic judgment about the student.

## 6. Loops repeat an analysis

We can apply the same rule to many values. `for` visits each item. The list below is a simple collection; we will study lists properly next week.

```python
scores = [78, 45, 62, 39, 91]
for score in scores:
    print(score)
```

A running total begins at zero and grows once per score. Predict the final value before running.

```python
total = 0
for score in scores:
    total = total + score
print("Total:", total)
print("Average:", total / len(scores))
```
