## 2. Data and the workflow

**Data** is recorded information. A score is data; so are the date of the quiz, whether a student was absent, and how the quiz was marked. Data is useful only when we connect it to a clear question.

| Step | Question to ask in our case |
|---|---|
| Define | Are we identifying students who may need support, or deciding who passes? |
| Collect | Where did scores come from? Do we have permission to use them? |
| Check and clean | Are scores missing, duplicated, or outside 0–100? |
| Explore | What is the average? How many are below 50? |
| Explain | What pattern did we find, and how uncertain is it? |
| Act and review | Offer support, then check whether it helped. |

Prediction or machine learning can come later, **if the question and data justify it**. It is not a required step in every data project.

## 7. Build: Student Grade Analyzer (15–18 minutes)

Use the five scores below. For each score, print its grade using the rules from section 5. At the end, print the class average and the number below 50. **Build it in stages:** first grades, then the count, then the average. Check that the average is `63.0` and the review count is `2`.

**Reasoning check:** If the school changes the review threshold to below 60, which result changes? Can you change one variable instead of several lines?

**Optional extension:** Use `names = ["Ada", "Tunde", "Chioma", "Emeka", "Zainab"]` and `zip(names, scores)` to include names. Do this only after the core result works.

```python
#Grade analyzer
scores = [78, 45, 62, 39, 91]
review_threshold = 50
# Stage 1: print a grade for every score.
for student_score in scores:
    if student_score < 50: 
        print("Needs review", student_score)
    elif student_score < 70:
        print("Pass", student_score)
    else:
        print("Excellent", student_score)
        
# Stage 2: count scores below review_threshold.
count = 0  
for student_score in scores:
    if student_score < review_threshold:
        count += 1
print("Number of scores below review threshold:", count)

# Stage 3: calculate and print the average.
total = sum(scores)
average = total/len(scores)
print("Average:", average)
```
