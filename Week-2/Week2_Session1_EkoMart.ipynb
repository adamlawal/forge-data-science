### 3.2 Mutability and the copy trap

Assigning a list to a new name does **not** copy it. Both names point to the same list. This is one of the most common beginner bugs.

````python
todays_orders = [101, 102, 103]
backup = todays_orders          # NOT a copy
backup.append(999)
print("todays_orders:", todays_orders)   # surprise: 999 is here too

todays_orders = [101, 102, 103]
real_backup = todays_orders.copy()
real_backup.append(999)
print("todays_orders:", todays_orders, "| real_backup:", real_backup)
````

### 4.2 Sets: the magic filter

A set holds **unique** items in **no particular order**. Use sets to remove duplicates and to compare groups.

| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `a \| b` | `a.union(b)` | In either group |

`print` shows a value on screen. `return` hands the value back so the rest of your program can use it. A function without `return` gives back `None`; this is the third common mistake in our course plan.

````python
def total_with_print(price, quantity):
    print(price * quantity)

def total_with_return(price, quantity):
    return price * quantity

a = total_with_print(4500, 2)
b = total_with_return(4500, 2)
print("a is", a, "| b is", b)
print("Can we add VAT to b?", b * 1.075)
````

### 6.2 Local and global variables

Variables created inside a function are **local**: they exist only while the function runs. Variables created outside are **global**. A function can read a global, but it is cleaner to pass values in as parameters.

````python
vat_rate = 0.075          # global

def add_vat(amount):
    vat = amount * vat_rate    # vat is local
    return amount + vat
````

````python
print(vat)
````

````python
print(add_vat(10000))

try:
    print(vat)
except NameError as e:
    print("Outside the function:", e)
````

## 7. Errors and exceptions

There are two broad kinds of error:

- **Syntax errors** (including `IndentationError`): Python cannot even read the code. Nothing runs until you fix it.
- **Runtime errors (exceptions)**: the code is valid, but something goes wrong while it runs.

| Exception | Typical cause | EkoMart example |
|---|---|---|
| `ZeroDivisionError` | Dividing by zero | Average order value on a day with 0 orders |
| `TypeError` | Wrong type for an operation | `"5000" + 1500` |
| `ValueError` | Right type, impossible value | `int("five")` |
| `KeyError` | Missing dictionary key | `product["colour"]` |
| `IndexError` | List position does not exist | `orders[10]` on a 7-item list |
| `NameError` | Name never defined, often a typo | `print(totl)` |
| `AttributeError` | Method does not exist for that type | `[1, 2].upper()` |

Read errors from the **bottom line up**: the last line names the exception and the reason; the lines above show where it happened.

````python
day_sales = 0
day_orders = 0
average = day_sales / day_orders
````

### 7.1 `try` and `except`

Wrap only the risky line, and catch the **specific** error you expect. A bare `except:` hides every bug, including typos you would want to see.

````python
day_sales, day_orders = 0, 0
try:
    average = day_sales / day_orders
except ZeroDivisionError:
    average = 0
    print("No orders today, so the average is set to 0")
print("Average:", average)
````

````python
raw_quantities = ["3", "two", "5", "", "1"]
clean = []
for q in raw_quantities:
    try:                                   # try/except INSIDE the loop: one bad value does not stop the rest
        number = int(q)
    except ValueError as error:
        print(f"Skipping {q!r}: {error}")  # printing the error message helps debugging
    else:
        clean.append(number)               # runs only if no error happened
    finally:
        pass                               # runs every time; used for clean-up such as closing files
print("Clean quantities:", clean)
````

### 7.2 Functions that handle their own errors

A well-built function protects its callers from crashing and returns something sensible.

````python
def safe_price(product, catalogue_lookup):
    try:
        return catalogue_lookup[product]
    except KeyError:
        print(f"Warning: {product} is not in the catalogue")
        return None

print(safe_price("P02", price_lookup))
print(safe_price("P99", price_lookup))
````

## 10. EkoMart Automated Sales Auditor

The manager wants a daily audit of orders that checks every order, flags problems and summarises the day. We will build it in four steps. 
##### Read through and understand the flow
### Step 1: The data

````python
prices = {"P01": 139000, "P02": 9800, "P03": 18500, "P04": 32000, "P05": 14400}

orders = [
    {"id": 1001, "customer": "Ada Okafor",     "area": "Ikeja",    "sku": "P02", "qty": 3,  "paid": 29400},
    {"id": 1002, "customer": "Tunde Bakare",   "area": "Lekki",    "sku": "P01", "qty": 1,  "paid": 139000},
    {"id": 1003, "customer": "Chioma Eze",     "area": "Yaba",     "sku": "P05", "qty": 2,  "paid": 25000},
    {"id": 1004, "customer": "Emeka Obi",      "area": "Ajah",     "sku": "P09", "qty": 1,  "paid": 5000},
    {"id": 1005, "customer": "Zainab Musa",    "area": "Surulere", "sku": "P04", "qty": 0,  "paid": 0},
    {"id": 1006, "customer": "Bola Ade",       "area": "Ikeja",    "sku": "P03", "qty": 2,  "paid": 37000},
    {"id": 1002, "customer": "Tunde Bakare",   "area": "Lekki",    "sku": "P01", "qty": 1,  "paid": 139000},
    {"id": 1007, "customer": "Kemi Lawal",     "area": "Lekki",    "sku": "P02", "qty": "two", "paid": 19600},
]
print(len(orders), "orders loaded")
````

### Step 2: One function to check one order

The function returns a list of problems, if a list is empty it means the order is clean.

````python
def audit_order(order, price_lookup):
    problems = []
    try:
        qty = int(order["qty"])
    except ValueError:
        return [f"quantity {order['qty']!r} is not a number"]

    if qty <= 0:
        problems.append("quantity must be at least 1")

    price = price_lookup.get(order["sku"])
    if price is None:
        problems.append(f"unknown product {order['sku']}")
    elif order["paid"] != price * qty:
        problems.append(f"paid ₦{order['paid']:,} but expected ₦{price * qty:,}")
    return problems

print(audit_order(orders[0], prices))
print(audit_order(orders[2], prices))
````

### Step 3: Audit every order, and catch duplicates with a set

````python
seen_ids = set()
clean_orders = []
flagged = {}

for order in orders:
    if order["id"] in seen_ids:
        flagged[order["id"]] = flagged.get(order["id"], []) + ["duplicate order id"]
        continue
    seen_ids.add(order["id"])

    problems = audit_order(order, prices)
    if problems:
        flagged[order["id"]] = problems
    else:
        clean_orders.append(order)

for order_id, problems in flagged.items():
    print(order_id, "->", "; ".join(problems))
````

### Step 4: Summarise the clean orders

````python
def summarise(clean):
    revenue = sum(o["paid"] for o in clean)
    by_area = {}
    for o in clean:
        by_area[o["area"]] = by_area.get(o["area"], 0) + o["paid"]
    return revenue, by_area

revenue, by_area = summarise(clean_orders)
print(f"Clean orders: {len(clean_orders)} of {len(orders)}")
print(f"Verified revenue: ₦{revenue:,}")
for area, amount in sorted(by_area.items(), key=lambda pair: pair[1], reverse=True):
    print(f"  {area:<9} ₦{amount:,}")
````

## 13. Recap

| Tool | Use it when | Key syntax |
|---|---|---|
| List | Order matters and items will change | `[ ]`, `append`, `pop`, slicing, `copy()` |
| Tuple | Values belong together and must not change | `( )`, `("one",)`, unpacking |
| Set | You need unique items or to compare groups | `{ }`, `&`, `\|`, `-`, `add` |
| Dictionary | You look things up by name | `{key: value}`, `get`, `items()` |
| List of dictionaries | You have a table of records | `[{...}, {...}]`, `sorted(key=...)` |
| Function | You repeat the same logic | `def`, parameters, defaults, `return` |
| try / except | A specific error is expected | `except ValueError`, `else`, `finally` |
| Module | Someone has already written it | `import`, `from ... import`, `as` |

Three mistakes to watch for this week: changing a list through a second name, using a position on a dictionary (or a key on a list), and writing a function that prints but does not return.
